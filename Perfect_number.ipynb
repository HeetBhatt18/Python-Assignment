# **Q15. Check perfect number.**

```python
def is_perfect(num):

    if num <= 0:
        return False

    divisor_sum = 0

    for i in range(1, num):
        if num % i == 0:
            divisor_sum = divisor_sum + i

    if divisor_sum == num:
        return True
    else:
        return False

num = int(input("Enter a number to check: "))

if is_perfect(num):
    print(f"{num} is a perfect number!")
else:
    print(f"{num} is NOT a perfect number.")
```
