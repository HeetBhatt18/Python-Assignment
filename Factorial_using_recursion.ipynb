# **Q10. Factorial using recursion.**

```python
def factorial(n):
  if n==0 or n==1:
    return 1

  else:
    return n * factorial(n-1)

num = int(input("Enter a number to find a factorial: "))

if num < 0:
  print("Factorial is not defined for negative numbers!")

else:
  result = factorial(num)
  print(f"{num}! = {result}")
```
